# Clean raw data

Select an archived raw-data snapshot by its ISO date. The cleaned output keeps the same date, so historical inputs and their derived datasets remain available.

```python
from pathlib import Path
import sys

def find_project_root(start: Path) -> Path:
    for candidate in (start, *start.parents):
        if (candidate / 'Data').is_dir() and (candidate / 'Scraper').is_dir():
            return candidate
    raise FileNotFoundError('Could not find the project root.')

PROJECT_ROOT = find_project_root(Path.cwd().resolve())
DATA_DIR = PROJECT_ROOT / 'Data'
SCRAPER_DIR = PROJECT_ROOT / 'Scraper'
RAW_DATA_DIR = DATA_DIR / 'raw'
CLEAN_DIR = DATA_DIR / 'clean'
SNAPSHOT_DATE = '2026-05-25'  # YYYY-MM-DD; choose the raw snapshot to clean
RAW_DATA_PATH = RAW_DATA_DIR / f'{SNAPSHOT_DATE}_used_cars_raw.csv'
CLEANED_DATA_PATH = CLEAN_DIR / f'{SNAPSHOT_DATE}_used_cars_cleaned.csv'

DATA_DIR.mkdir(parents=True, exist_ok=True)
RAW_DATA_DIR.mkdir(parents=True, exist_ok=True)
CLEAN_DIR.mkdir(parents=True, exist_ok=True)

if str(SCRAPER_DIR) not in sys.path:
    sys.path.insert(0, str(SCRAPER_DIR))

from clean_used_cars_csv import clean_cars
```

```python
cleaned_cars = clean_cars(RAW_DATA_PATH, CLEANED_DATA_PATH)

print(f'Raw snapshot: {RAW_DATA_PATH}')
print(f'Clean ML dataset saved to: {CLEANED_DATA_PATH}')
print(f'Unique listings: {len(cleaned_cars):,}')
cleaned_cars.head()
```
