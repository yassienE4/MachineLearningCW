# Used-car data collection

Run the setup, collection, and cleaning cells in order to scrape Hatla2ee listings and create a date-versioned, de-duplicated dataset. Each collection snapshot is retained under `Data/raw/` and `Data/clean/`.

```python
from datetime import date
from pathlib import Path
import sys

def find_project_root(start: Path) -> Path:
    for candidate in (start, *start.parents):
        if (candidate / 'Data').is_dir() and (candidate / 'Scraper').is_dir():
            return candidate
    raise FileNotFoundError('Could not find the project root.')

PROJECT_ROOT = find_project_root(Path.cwd().resolve())
SCRAPER_DIR = PROJECT_ROOT / 'Scraper'
DATA_DIR = PROJECT_ROOT / 'Data'
RAW_DATA_DIR = DATA_DIR / 'raw'
CLEAN_DIR = DATA_DIR / 'clean'
SNAPSHOT_DATE = date.today().isoformat()
CLEANED_DATA_PATH = CLEAN_DIR / f'{SNAPSHOT_DATE}_used_cars_cleaned.csv'

DATA_DIR.mkdir(parents=True, exist_ok=True)
RAW_DATA_DIR.mkdir(parents=True, exist_ok=True)
CLEAN_DIR.mkdir(parents=True, exist_ok=True)

if str(SCRAPER_DIR) not in sys.path:
    sys.path.insert(0, str(SCRAPER_DIR))

from hatla2ee_scraper import scrape_cars
from clean_used_cars_csv import clean_cars
```

```python
raw_data_path = scrape_cars(
    output_dir=str(RAW_DATA_DIR),
    snapshot_date=SNAPSHOT_DATE,
)

print(f'Raw scrape saved to: {raw_data_path}')
```

```python
# This cell can run without first running the scrape cell.
raw_data_path = Path('/Users/yassien/Documents/Uni/MachineLearning/Project/Data/raw/2026-10-13_used_cars_raw.csv')
snapshot_date = raw_data_path.name.removesuffix('_used_cars_raw.csv')
CLEANED_DATA_PATH = CLEAN_DIR / f'{snapshot_date}_used_cars_cleaned.csv'

if not raw_data_path.is_file():
    raise FileNotFoundError(f'Raw dataset not found: {raw_data_path}')

cleaned_cars = clean_cars(raw_data_path, CLEANED_DATA_PATH)

print(f'Clean ML dataset saved to: {CLEANED_DATA_PATH}')
print(f'Unique listings: {len(cleaned_cars):,}')
cleaned_cars.head()
```
